## Converting Date/Time Columns

The dataset contains four columns representing timestamps:

- `trip_creation_time`
- `od_start_time`
- `od_end_time`
- `cutoff_timestamp`

These columns are initially stored as `object` (text) data types.

For time-based analysis, they need to be converted to Pandas `datetime` format.

```python
date_columns = [
    "trip_creation_time",
    "od_start_time",
    "od_end_time",
    "cutoff_timestamp"
]

for col in date_columns:
    df_clean[col] = pd.to_datetime(df_clean[col], errors="coerce")
```

### Verify Datetime Conversion

We verify the data types after conversion to ensure that all timestamp columns are correctly recognized as datetime values.

```python
df_clean[date_columns].dtypes
```

### Check for Invalid or Unconverted Timestamps

The conversion uses `errors="coerce"`, which converts any invalid timestamp values into `NaT` (Not a Time).

## Handling Missing Cutoff Timestamps

After converting `cutoff_timestamp` to datetime format, 3,429 records were converted to `NaT`.

The affected records include both:

- 2,841 records where `is_cutoff = False`
- 588 records where `is_cutoff = True`

Because the original timestamp information cannot be reliably reconstructed from the available dataset, these values are retained as `NaT` rather than being replaced with an arbitrary timestamp.

The associated `is_cutoff` and `cutoff_factor` fields are preserved because they still contain useful operational information.

This approach avoids introducing fabricated data while retaining the affected logistics records for further analysis.

```python
cutoff_summary = pd.DataFrame({
    "Metric": [
        "Total records",
        "Missing cutoff timestamps",
        "Missing cutoff timestamps (%)"
    ],
    "Value": [
        len(df_clean),
        df_clean["cutoff_timestamp"].isna().sum(),
        round(df_clean["cutoff_timestamp"].isna().mean() * 100, 2)
    ]
})

cutoff_summary
```

### Dataset Granularity

The dataset contains 144,867 records but only 14,817 unique trips identified by `trip_uuid`.

This indicates that multiple records can belong to the same trip. Therefore, the dataset contains information at a more granular level than the overall trip.

The dataset also contains 1,508 unique source centers and 1,481 unique destination centers.

For analysis, the appropriate level of aggregation will depend on the business question. Segment-level analysis will be used for detailed route performance, while `trip_uuid` will be used to aggregate records when evaluating overall trip performance.
